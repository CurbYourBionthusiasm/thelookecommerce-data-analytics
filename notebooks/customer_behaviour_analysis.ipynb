```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../data/orders_per_customer.csv')
```

```python
order_distribution = (df['orders_per_customer'].value_counts(normalize = True).sort_index()*100)

print(f"""1-time customer percentage: {round(order_distribution[1],2)}%,
2-time customer percentage: {round(order_distribution[2],2)}%,
3-time customer percentage: {round(order_distribution[3],2)}%,
4-time customer percentage: {round(order_distribution[4],2)}%
""")
```

```python
repeat_customers = (df['orders_per_customer']>1).sum()
total_customers = len(df)

repeat_rate = round((repeat_customers/total_customers)*100,2)

print(f"Repeat purchase rate: {repeat_rate}%")
```

```python
plt.figure(figsize=(8,5))

sns.countplot(x = df['orders_per_customer'])
plt.title('Distribution of orders per customer')
plt.xlabel('Number or orders')
plt.xticks(ticks=(1,2,3,4))
plt.ylabel('Number of customers')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.barplot(x = order_distribution.index, y = order_distribution.values)
plt.title('Percentage distribution of orders per customer')
plt.xlabel('Number of orders')
plt.ylabel('Percentage of customers')

plt.show()
```

#Key insights:

Most customers place only a single order (88,03\%).

Repeat purchases represent a smaller portion of the customer base (11,97\%) ,
indicating that customer retention could be an important driver
of future revenue growth.

#Business recommendation:

The business could focus on improving customer retention.

Encouraging repeat purchases through loyalty programs,
personalized offers, or post-purchase engagement
could significantly increase long-term revenue.
