```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Load data
df = pd.read_csv("..\\data\\revenue_per_order.csv")
```

```python
#Basic statistics
mean = round(df['revenue_per_order'].mean(),2)
median = round(df['revenue_per_order'].median(),2)
minimal = df['revenue_per_order'].min()
maximal = df['revenue_per_order'].max()

orders_above_500 = (df["revenue_per_order"] >= 500).sum()
share_above_500 = round(orders_above_500/len(df)*100,2)


print(f"""Mean order value = ${mean}, 
median order value = ${median}, 
minimal order value = ${minimal},
maximal order value = ${maximal}.

There are {orders_above_500} orders which value exceeds $500. Which is {share_above_500}% of all orders.""")
```

```python
#Histplot
plt.figure(figsize=(8,5))
sns.histplot(df["revenue_per_order"],bins=50,kde=True)


plt.title("Distribution of Order Value")
plt.xlabel("Order Value ($)")
plt.ylabel("Number of Orders")

plt.show()
```

##Key insights:

The distribution of order value is right-skewed.

While the mean order value is \\$86.21 , the median value
is significantly lower at \\$55.5, indicating that a small
number of high-value orders increase the average.

Orders above \\$500 represent only a small fraction
of all transactions.

##Business implication:

The presence of a long tail suggests that average order value may not
fully represent typical customer purchases. Median order value may
provide a better indicator of a typical transaction size.
